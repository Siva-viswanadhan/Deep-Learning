```python
count=0
for i in range(n):
    if dic3['conf'][i] >=50:
        x=dic3['left'][i]
        y=dic3['top'][i]
        w=dic3['width'][i]
        h=dic3['height'][i]
        if dic3['text'][i] =='dog':
            count+=1
            cv2.rectangle(img3,(x,y),(x+w,y+h),color=[0,0,255],thickness=1)
    

cv2.imshow('text',img3)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```
