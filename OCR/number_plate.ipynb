```python
import cv2
import pytesseract
import re
```

```python
pytesseract.pytesseract.tesseract_cmd=r'C:\Program Files\Tesseract-OCR\tesseract.exe'
```

```python
img=cv2.imread(r"D:\vs_code_DL\files\WhatsApp Image 2025-12-02 at 13.17.26_5df14fca.jpg")
```

```python
# 1. Convert to gray
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 2. Denoise
gray = cv2.bilateralFilter(gray, 11, 17, 17)

# 3. Threshold (important)
_, img_binary = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
```

```python
data=pytesseract.image_to_data(img_binary,output_type=pytesseract.Output.DICT)
```

```python
for i in range(len(data['text'])):
    if re.search(r'^[A-Z]{2}[ -]?[0-9]{1,2}[ -]?[A-Z]{1,3}[ -]?[0-9]{1,4}$',data['text'][i]) or re.search(r'^[0-9]{2}[ -]?BH[ -]?[0-9]{4}[ -]?[A-Z]{2}$',data['text'][i]):
        print("Detected:", data['text'][i])
        x = data['left'][i]
        y = data['top'][i]
        w = data['width'][i]
        h = data['height'][i]
        cv2.rectangle(img,(x,y),(x+w,y+h),color=(0,0,255),thickness=2)
cv2.imshow("nb",img)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```
