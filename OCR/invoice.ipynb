```python
import cv2
import pytesseract
```

```python
pytesseract.pytesseract.tesseract_cmd=r'C:\Program Files\Tesseract-OCR\tesseract.exe'
```

```python
img2=cv2.imread(r'D:\vs_code_DL\files\ocr2.jpeg.jpg')
```

```python
text2=pytesseract.image_to_string(img2)
print(text2)
```

```python
dic2=pytesseract.image_to_data(img2,output_type=pytesseract.Output.DICT)
```

```python
n=len(dic2['text'])
n
```

```python
for i in range(n):
    if dic2['conf'][i] >= 50:
        x=dic2['left'][i]
        y=dic2['top'][i]
        w=dic2['width'][i]
        h=dic2['height'][i]

        cv2.rectangle(img2,(x,y),(x+w,y+h),(0,0,255,3))

cv2.imshow('img',img2)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
import re
for i in range(n):
    if re.search(r'(\d{1,2}/\d{1,2}/\d{4})',dic2['text'][i]):
        print(dic2['text'][i])
```
