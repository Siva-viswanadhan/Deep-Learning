```python
count = 0

for skill in skills_list:
    found = 0
    for text in dic['text']:
        if skill.lower() in text.lower():
            found = 1
            break
    if found ==1:
        count += 1

print(count)
```

```python
percentage=(count/len(skills_list))*100
percentage
```
