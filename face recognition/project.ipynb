```python
def face_encoding(images):
    encode_list=[]

    for i in images:
        img=cv2.cvtColor(i,cv2.COLOR_BGR2RGB)
        face_in_frame=fr.face_locations(img)
        # print(face_in_frame)
        encoder=fr.face_encodings(img,face_in_frame)[0]
        encode_list.append(encoder)
        # print(encode_list)
    return encode_list

```
