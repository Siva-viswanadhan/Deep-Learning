```python
import ultralytics
```

```python
model=ultralytics.YOLO('yolov8n.pt')
import cv2
import matplotlib.pyplot as plt

result=model([r'D:\vs_code_DL\files\street.jpg'])


for i in result:
    i.save(filename='result.jpg')
    img=cv2.imread('result.jpg')
    img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
    plt.imshow(img)
```
