```python
import ultralytics
```

```python
model=ultralytics.YOLO('yolov8n.pt')
import cv2
```

```python
video = cv2.VideoCapture(0)

while True:
    suc, img = video.read()
    if not suc:
        break

    results = model(img)

    for r in results:
        r.save(filename='result1.jpg')
        img=cv2.imread('result1.jpg')
        cv2.imshow('image',img) 

    

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

video.release()
cv2.destroyAllWindows()
```
