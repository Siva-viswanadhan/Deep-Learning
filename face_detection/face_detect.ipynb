```python
import cv2
import matplotlib.pyplot as plt
```

```python
img1=cv2.imread(r'D:\vs_code_DL\face_detection\360_F_326085309_CFH8PpadfnL2OQ7Gi411XW0B21YumxKo.jpg')
```

```python
con_img1=cv2.cvtColor(img1,cv2.COLOR_BGR2RGB)
plt.imshow(con_img1)
```

reading the model

```python
face_data=cv2.CascadeClassifier(r'D:\vs_code_DL\face_detection\haarcascade_frontalface_default.xml')
face_data
```

```python
faces=face_data.detectMultiScale(con_img1,minNeighbors=5)#minneigbors 5 only detect if detect 5 times
faces
```

first two values left top points and last two points height and width(not right bottom)


```python
for (x,y,w,h) in faces:
    cv2.rectangle(con_img1,pt1=(x,y),pt2=(x+w,y+h),color=[255,0,0],thickness=3)
```

```python
plt.imshow(con_img1)
```
