```python
import cv2
```

```python
face_detect_model=cv2.CascadeClassifier(r'D:\vs_code_DL\face_detection\haarcascade_frontalface_default.xml')
```

```python
#video

video=cv2.VideoCapture(0)

#read
while True:
    sucess,img=video.read()
    gray_img=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
    faces=face_detect_model.detectMultiScale(gray_img,minNeighbors=20)
    # print(faces)
    for (x,y,w,h) in faces:
        cv2.rectangle(img,(x,y),(x+w,y+h),color=[0,255,0],thickness=1) # img in bgr so[0,0,255] will be blue color


    cv2.imshow('video',img)
    if cv2.waitKey(1) & 0XFF==ord('q'): #hexadecimal == q ,q when pressed the window will close
        break


video.release()   # closing video
cv2.destroyAllWindows()



```
